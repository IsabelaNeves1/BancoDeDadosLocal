```python
# Instalação e download de recursos

!pip install -q spacy pandas
!python -m spacy download pt_core_news_sm -q

import spacy
import pandas as pd

nlp = spacy.load("pt_core_news_sm")

print("Ambiente configurado com sucesso.")
```

```python
frase_exercicio = "Olá! Gostaria de agendar uma visita para ver o imóvel de 3 quartos amanhã às 14h?"

# 1. Tokenização com spaCy
doc_spacy = nlp(frase_exercicio)
tokens_spacy = [token.text for token in doc_spacy]
print("Tokens com spaCy:  ", tokens_spacy)
```

```python
# 2. Remoção de pontuação (use token.is_punct)

def remover_pontuacao(frase_exercicio):
    doc = nlp(frase_exercicio)
    retirar_pontuacao = [token.text for token in doc if not token.is_punct]
    return (retirar_pontuacao)
frase_exercicio_sp = remover_pontuacao(frase_exercicio)
```

```python
print(frase_exercicio_sp)
```

```python
# 3. Contagem: original vs. sem pontuação
print(f"Quantidade de Tokens Original: {len(tokens_spacy)} \n Quantidade de Tokens Sem pontuação: {len(frase_exercicio_sp)}")
```
