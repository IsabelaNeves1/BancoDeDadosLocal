```python
# Instalação e download de recursos (rodar 1x)
!pip install -q spacy pandas
!python -m spacy download pt_core_news_sm -q

import spacy
import pandas as pd
from spacy import displacy

nlp = spacy.load("pt_core_news_sm")

print("Ambiente configurado com sucesso.")
```

```python
frase_ancora = "O banco aprovou o empréstimo ontem."
doc_ancora = nlp(frase_ancora)

print([token.text for token in doc_ancora])
```

```python
displacy.render(doc_ancora, style="dep", jupyter=True, options={"distance": 100, "collapse_punct": False})
```

```python
raiz = [token for token in doc_ancora if token.dep_ == "ROOT"][0]
print("Raiz da frase (verbo principal):", raiz.text)

# Encontrar o sujeito (nsubj = nominal subject)
sujeito = [token for token in doc_ancora if token.dep_ == "nsubj"][0]
print("Sujeito:", sujeito.text)

# Sintagma Nominal (SN): o sujeito e tudo que depende dele (ex.: o artigo "O")
sintagma_nominal = list(sujeito.subtree)
print("Sintagma Nominal (SN):", [t.text for t in sintagma_nominal])

# Sintagma Verbal (SV): a raiz e tudo que depende dela, exceto o próprio SN
sintagma_verbal = [t for t in raiz.subtree if t not in sintagma_nominal]
print("Sintagma Verbal (SV):", [t.text for t in sintagma_verbal])
```

```python
frase_ambigua = "Vi o menino com o telescópio."
doc_ambiguo = nlp(frase_ambigua)

displacy.render(doc_ambiguo, style="dep", jupyter=True, options={"distance": 100, "collapse_punct": False})
```
