```python
import pandas as pd
import spacy
from spacy import displacy
```

```python
try:
    nlp = spacy.load("pt_core_news_sm")
except OSError:
    !python -m spacy download pt_core_news_sm
    nlp = spacy.load("pt_core_news_sm")
```

```python
frase_exercicio = "O corretor apresentou o imóvel para o cliente ontem."
doc_exercicio = nlp(frase_exercicio)
```

```python
# 1. POS Tagging
dados_pos = [
    {
        "Token": token.text,
        "Lema": token.lemma_,
        "Classe (POS)": token.pos_,
        "Tag Detalhada": token.tag_,
        "Explicação": spacy.explain(token.pos_),
    }
    for token in doc_exercicio
]

df_pos = pd.DataFrame(dados_pos)
display(df_pos)
```

```python
# 2. Visualização com displacy
displacy.render(
    doc_exercicio,
    style="dep",
    jupyter=True,
    options={"distance": 100, "collapse_punct": False},
)
```

```python
# 3. Extração do SN e do SV
raiz = [token for token in doc_exercicio if token.dep_ == "ROOT"][0]
sujeito = [token for token in doc_exercicio if token.dep_ == "nsubj"][0]
```

```python
# Sintagma Nominal (SN): sujeito e dependentes
sintagma_nominal = list(sujeito.subtree)
```

```python
# Sintagma Verbal (SV): predicado (raiz e dependentes, excluindo o SN)
sintagma_verbal = [token for token in raiz.subtree if token not in sintagma_nominal]
```

```python
print("Raiz:", raiz.text)
print("Sujeito:", sujeito.text)
print("Sintagma Nominal (SN):", [t.text for t in sintagma_nominal])
print("Sintagma Verbal (SV):", [t.text for t in sintagma_verbal])
```
